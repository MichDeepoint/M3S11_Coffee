Разделим выборку в соотношении 80/20, где 80 — обучающая и валидационные выборки, 20 — тестовая выборка.

````python
# Функция разделения данных
def separate_data(df, test_size, random_state, target_column, drop_columns=[], shuffle=True):
    # Признаки. Исключаем целевую переменную, мультиколенианые признаки и неважные признаки
    if drop_columns:
        drop_columns.append(target_column)
        X = df.drop(columns=drop_columns).copy()

    # Целевая переменная
    y = df[target_column].copy()

    # Разделение 
    X_train_val, X_test, y_train_val, y_test = train_test_split(X, y, 
                                                                test_size=test_size, 
                                                                random_state=random_state,
                                                                stratify=y, 
                                                                shuffle=shuffle)  
    
    return X_train_val, X_test, y_train_val, y_test 

# список искючаемых столбцов из выборки признаков
drop_columns = ['user_id', 'order_frequency_week', 'median_order_value']

# Целевая переменная
target = 'churn'

X_train, X_test, y_train, y_test  = separate_data(df, test_size=0.2, random_state=RANDOM_STATE,
                                                  target_column=target, drop_columns=drop_columns)

print(f"Размер обучающей выборки: {X_train.shape}")
print(f"Размер тестовой выборки: {X_test.shape}")
print(f"Доля положительного класса в train+val: {y_train.mean():.3f}")
print(f"Доля положительного класса в test: {y_test.mean():.3f}")
````

### <a id='toc2_3_2_'></a>[Пайплайн по обработке пропусков, отрицательных значений и выбросов](#toc0_)

Создадим пайплайн для обработки пропусков, отрицательных значений и выбросов.

````python
# Функция заполнения нулём отрицательных значений
def replace_negatives_with_zero(X):    
        return X.clip(lower=0)

# Функция создания пайплайна по заполнению пропусков, замене отрицательных значений и логарифмированию выбросов
def create_imputation_pipeline(numeric_zero_fill=None, numeric_big_value_fill=None, numeric_median_fill=None,
                               categorical_unknow_fill=None, categorical_none_fill=None, 
                               categorical_mode_fill=None, negative_cols=None, outlier_cols=None):
               
    # Числовые признаки, заполнение нулём
    if numeric_zero_fill is None:
        numeric_zero_fill = [
            'order_frequency_month', 'total_spent_last_month', 
            'total_spent_last_week','app_opens_per_week', 
            'app_crashes_last_month', 'discount_usage_rate', 
            'seasonal_menu_tried', 'notifications_enabled', 
            'coffee_preference_change']

    # Числовые признаки, заполнение отрицательныз значений
    if negative_cols is None:
        negative_cols = ['avg_order_value', 'total_spent_last_month', 
                         'total_spent_last_week', 'review_rating_last_1']

    # Числовые признаки, заполнение большим значением
    if numeric_big_value_fill is None:
        numeric_big_value_fill = ['days_since_last_order', 'days_since_last_promo']

    # Числовые признаки, заполнение медианой
    if numeric_median_fill is None:
        numeric_median_fill = ['avg_order_value', 'review_rating_last_10', 'review_rating_last_1']    
    
    # Категориальные признаки, заполнение Unknown
    if categorical_unknow_fill is None:
        categorical_unknow_fill = [
            'last_coffee_type', 'preferred_roast', 'milk_preference',
            'coffee_bean_origin', 'last_drink_size', 'phone_type',
            'geo_location']

    # Категориальные признаки, заполнение none
    if categorical_none_fill is None:
        categorical_none_fill = ['subscription_status']

    # Категориальные признаки, заполнение модой
    if categorical_mode_fill is None:
        categorical_mode_fill = ['seasons']    

    # Выбросы
    if outlier_cols is None:
        outlier_cols = ['days_since_last_order', 'total_spent_last_month', 
                        'total_spent_last_week', 'app_opens_per_week', 'days_since_last_promo']
          
    
    # Создаем импутеры для числовых признаков
    numeric_zero_transformer = SimpleImputer(strategy='constant', fill_value=0)
    numeric_big_value_transformer = SimpleImputer(strategy='constant', fill_value=999)
    numeric_median_transformer = SimpleImputer(strategy='median')    

    # Создаём импутеры для категориальных признаков
    categorical_unknow_transformer = SimpleImputer(strategy='constant', fill_value='unknown')    
    categorical_none_transformer = SimpleImputer(strategy='constant', fill_value='none')
    categorical_mode_transformer = SimpleImputer(strategy='most_frequent')

    # Заполнение отрицательных значений
    numeric_fill_neg = FunctionTransformer(replace_negatives_with_zero)

    # Логарифмирование для выбросов
    numeric_log_transformer =  FunctionTransformer(np.log1p)
    
    imp_transformers = []
    if numeric_zero_fill:
        imp_transformers.append(('num_zero', numeric_zero_transformer, numeric_zero_fill))
    if numeric_big_value_fill:
        imp_transformers.append(('num_big_value', numeric_big_value_transformer, numeric_big_value_fill))
    if numeric_median_fill:
        imp_transformers.append(('num_median', numeric_median_transformer, numeric_median_fill))
    if categorical_unknow_fill:
        imp_transformers.append(('cat_unknown', categorical_unknow_transformer, categorical_unknow_fill))
    if categorical_none_fill:
        imp_transformers.append(('cat_non', categorical_none_transformer, categorical_none_fill))
    if categorical_mode_fill:
        imp_transformers.append(('cat_mode', categorical_mode_transformer, categorical_mode_fill))

    # CT заполнения пропусков
    ct_imp = ColumnTransformer(
        transformers=imp_transformers,
        remainder='passthrough',
        verbose_feature_names_out=False
    )
    ct_imp.set_output(transform='pandas')

    # CT замены отрицательных значений
    neg_transformers = []
    if negative_cols:
        neg_transformers.append(('fill_neg', numeric_fill_neg, negative_cols))
    
    if neg_transformers:
        ct_neg = ColumnTransformer(
            transformers=neg_transformers,
            remainder='passthrough',
            verbose_feature_names_out=False
        )
    else:
        ct_neg = 'passthrough'  
    ct_neg.set_output(transform='pandas') if hasattr(ct_neg, 'set_output') else None

    # CT логорифмирования
    log_transformers = []
    if outlier_cols:
        log_transformers.append(('num_log', numeric_log_transformer, outlier_cols))

    if log_transformers:
        ct_log = ColumnTransformer(
            transformers=log_transformers,
            remainder='passthrough',
            verbose_feature_names_out=False
        )
    else:
        ct_log = 'passthrough'
    ct_log.set_output(transform='pandas') if hasattr(ct_log, 'set_output') else None

    # Пайплайн
    preprocessor = Pipeline(steps=[
        ('imputation', ct_imp),
        ('fill_neg', ct_neg),
        ('num_log', ct_log)
    ])
    preprocessor.set_output(transform='pandas')
    return preprocessor 

# X_after_imp = create_imputation_pipeline().fit_transform(X_train)
# print(X_after_imp.isna().sum())
````

### <a id='toc2_3_3_'></a>[Пайплайн по обработке категориальных признаков](#toc0_)

Создадим функцию которая будет возвращать пайплайн по обработке категориальных признаков.

````python
# Функция кодирования категориальных значений
def create_encoding_pipeline(ohe_columns=None, te_columns=None):
    # Признаки для OHE
    if ohe_columns is None:
        ohe_columns = ['last_coffee_type','preferred_roast',  
                    'milk_preference','coffee_bean_origin',  
                    'last_drink_size', 'subscription_status', 
                    'seasons', 'phone_type']
    
    # Признаки для TE
    if te_columns is None:
        te_columns = ['geo_location']

    enc_transformers = []
    if ohe_columns:
        enc_transformers.append(('ohe_encoder', OneHotEncoder(sparse_output=False), ohe_columns))
    if te_columns:
        enc_transformers.append(('te_encoder', TargetEncoder(cv=5, smooth='auto', shuffle=True, random_state=RANDOM_STATE), te_columns))

    # CT заполнения пропусков
    if enc_transformers:
        ct_enc = ColumnTransformer(
            transformers=enc_transformers,
            remainder='passthrough',
            verbose_feature_names_out=False
        )
        ct_enc.set_output(transform='pandas')
    else:
        ct_enc = 'passthrough'    

    return ct_enc

# X_after_enc = create_encoding_pipeline().fit_transform(X_train, y_train)
# print(X_after_enc.isna().sum())
````

### <a id='toc2_3_4_'></a>[Пайплайн по масштабированию и нормализации числовых признаков](#toc0_)

Создадим пайплайн для масштабирования и нормализации числовых признаков.

````python
# Функция масштабирования и нормализации
def create_scaling_pipeline(std_scaler_columns=None, minmax_scaler_columns=None):

    # Признаки для масштабирования
    if std_scaler_columns is None:
        std_scaler_columns = ['avg_order_value', 'total_spent_last_month', 
                              'total_spent_last_week', 'review_rating_last_10',
                              'app_crashes_last_month', 'days_since_last_order',
                              'days_since_last_promo', 'app_opens_per_week', 
                              'order_frequency_month', 'discount_usage_rate']

    # Признаки для нормализации
    if minmax_scaler_columns is None:
        minmax_scaler_columns = ['review_rating_last_1']

    scal_transformers = []
    if std_scaler_columns:
        scal_transformers.append(('std_encoding', StandardScaler(), std_scaler_columns))
    if minmax_scaler_columns:
        scal_transformers.append(('minmax_encoding', MinMaxScaler(), minmax_scaler_columns))

    if scal_transformers:
        ct_scal = ColumnTransformer(
            transformers=scal_transformers,
            remainder='passthrough',
            verbose_feature_names_out=False
        )
        ct_scal.set_output(transform='pandas')
    else:
        ct_scal = 'passthrough'        
    
    return ct_scal    
````

<div class="alert alert-warning">

**Совет: пайплайн стоит собрать явно, а не через функции-фабрики 💡**

Замысел понятен — сделать создание пайплайнов параметризуемым и переиспользуемым (`create_imputation_pipeline`, `create_encoding_pipeline`, `create_scaling_pipeline` с настраиваемыми списками колонок). Это инженерный паттерн «фабрика», и в большой продакшн-кодовой базе он бывает уместен. Но для этой задачи он скорее мешает, чем помогает, и вот почему:

1. **Читаемость и «явное лучше неявного».** Чтобы понять, какой признак куда попадёт, приходится проваливаться внутрь функции и разбирать `if not ...: ... = [...]`. При явном `ColumnTransformer` вся карта «колонка → преобразование» видна сразу, в одном месте.

2. **Дублирование — источник ошибок.** Списки колонок заданы и в дефолтах фабрик, и отдельно в `create_features`. Их легко рассинхронизировать — что здесь и произошло: часть числовых признаков осталась без масштабирования (см. красный комментарий про StandardScaler). Единый список констант в начале ноутбука такую рассинхронизацию исключает.

3. **Мутабельные аргументы по умолчанию.** `def create_imputation_pipeline(numeric_zero_fill=[], ...)` — известный питоновский подводный камень: один и тот же список разделяется между вызовами. Сейчас спасает проверка `if not ...`, но сам паттерн лучше не заводить в привычку.

Стандартный подход, который как раз разбирали в теории, — один `ColumnTransformer` с под-пайплайнами по типам признаков, а списки колонок — явными константами сверху:

```python
NUM_COLS = [...]
CAT_COLS = [...]

num_pipe = Pipeline([("imputer", ...), ("scaler", ...)])
cat_pipe = Pipeline([("imputer", ...), ("ohe", ...)])

preprocess = ColumnTransformer([
    ("num", num_pipe, NUM_COLS),
    ("cat", cat_pipe, CAT_COLS),
])
```

Столько же строк, но каждый шаг виден с первого взгляда, и всё преобразование лежит в одном объекте. Это не ошибка — код рабочий; но на будущее «явное» здесь заметно выигрывает у «универсального».
</div>

<div class="alert alert-danger">

**Масштабируется только часть числовых признаков 🚩**

Пайплайн масштабирования охватывает лишь `avg_order_value`, `total_spent_last_month`, `total_spent_last_week` (StandardScaler) и `review_rating_last_1` (MinMaxScaler). Остальные непрерывные числовые признаки уходят через `remainder='passthrough'` **без масштабирования** - в том числе `app_crashes_last_month` (самый сильный предиктор, phik 0.85 с оттоком), `days_since_last_order` (пропуски заполняются значением 999!), `days_since_last_promo`, `app_opens_per_week`, `order_frequency_month`, `discount_usage_rate`, `review_rating_last_10`.

Для логистической регрессии это критично - и особенно потому, что дальше ты подбираешь регуляризацию (`C` и `L1`/`L2`):

1. **Смещение регуляризации.** Штраф применяется к коэффициентам по модулю. Признак с большим диапазоном (999 у `days_since_last_order`) обходится маленьким коэффициентом и почти не штрафуется, а признак из `[0, 1]` (`discount_usage_rate`) вынужден иметь большой коэффициент - и штрафуется несправедливо. При L1 (которую ты и выбрал, C=0.5) это значит, что зануляются не самые бесполезные признаки, а те, у кого просто мельче масштаб. То есть весь подбор гиперпараметров идёт на смещённой основе.
2. **Один признак в двух шкалах.** `app_crashes_last_month` остаётся немасштабированным, а его сгенерированный `_squared` - масштабируется. Модель видит один и тот же признак в двух несопоставимых диапазонах.

**Что исправить:** масштабировать StandardScaler'ом **все** непрерывные числовые признаки, а не выбранную вручную тройку. Проще всего собрать числовой поток так, чтобы scaler применялся ко всему числовому списку:

```python
num_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy=...)),
    ("scaler", StandardScaler()),
])
preprocess = ColumnTransformer([
    ("num", num_pipe, num_cols),   # ВСЕ непрерывные числовые
    ("cat", cat_pipe, cat_cols),
])
```

Бинарные 0/1 и колонки после OHE масштабировать не нужно - у них уже сопоставимый масштаб.

### <a id='toc2_4_1_'></a>[Обучение базовой модели](#toc0_)

Для сравнения целевой модели LogisticRegression с моделью-случайного угадывания, создадим и применим кросс-валидацию к базовой модели DummyClassifier, со стратегией 'most_frequent', которая подходит для сильно несбалансированных классов в целевой веременной, как в нашем случае. Так же создадим и применим кроссвалидацию к модели LogisticRegression. Сравним результаты.

````python
# Функция вывода метрик двух моделей с метриками PR AUC и F1 и их сравнения
def metrics_comparison_output(baseline_result, model_result, baseline_name='', model_name=''):    
    print('=' * 50)
    print('Результаты кросс-валидации:')
    print('=' * 50)

    print(f"Базовая модель ({baseline_name}):")
    for key, values in baseline_result.items():
        if key.startswith('test_'):
            metric_name = key.replace('test_', '')
            print(f"  {metric_name}: {values.mean():.4f} +/- {values.std():.4f}")
    print("-" * 30)

    print(f"Целевая модель train ({model_name}):")
    for key, values in model_result.items():
        if key.startswith('test_'):
            metric_name = key.replace('test_', '')
            print(f"  {metric_name}: {values.mean():.4f} +/- {values.std():.4f}")
    print("-" * 30)

    print('Улучшение:')
    dummy_prauc = baseline_result['test_average_precision'].mean()
    logreg_prauc = model_result['test_average_precision'].mean()
    print(f"  PR-AUC: {logreg_prauc - dummy_prauc:+.4f}")


    dummy_f1 = baseline_result['test_f1'].mean()
    logreg_f1 = model_result['test_f1'].mean()
    print(f"  F1:     {logreg_f1 - dummy_f1:+.4f}")
    print('=' * 50)
````

`average_precision` (PR AUC) как основная метрика - точный выбор для задачи с дисбалансом. В отличие от ROC AUC, который остаётся «внешне красивым» даже при слабом предсказании редкого класса, PR AUC чувствителен именно к качеству предсказания оттока. Это метрика из ТЗ.

````python
# Отключаем предупреждения (устаревание методов)
warnings.filterwarnings('ignore')

# Количество фолдов
N_FOLDS = 5

# Метрики
scoring = ['average_precision', 'f1']

# Настройка кросс-валидации, стратегия
kfold = StratifiedKFold(n_splits=N_FOLDS, shuffle=True, random_state=RANDOM_STATE)

# Создание базовой модели
pipeline_dummy = Pipeline(
    steps=[
        ('imputation', create_imputation_pipeline()),        
        ('scaling', create_scaling_pipeline()),
        ('encoding', create_encoding_pipeline()),
        ('baseline', DummyClassifier(strategy='most_frequent', random_state=RANDOM_STATE))
    ]
)

# Создание целевой модели
pipeline_logreg = Pipeline(
    steps=[
        ('imputation', create_imputation_pipeline()),        
        ('scaling', create_scaling_pipeline()),
        ('encoding', create_encoding_pipeline()),
        ('logreg', LogisticRegression(max_iter=1000, random_state=RANDOM_STATE))
    ]
)

# Кросс-валидация модели dummy
dummy_cv_results = cross_validate(pipeline_dummy, X_train, y_train, cv=kfold, scoring=scoring)

# Кросс-валидация целевой модели
logreg_cv_results = cross_validate(pipeline_logreg, X_train, y_train, cv=kfold, scoring=scoring)

# Вывод сравнения
metrics_comparison_output(dummy_cv_results, logreg_cv_results, 'DummyClassifier', 'LogisticRegression')
````

````python
# Функция создания новых признаков
def create_features(X):

    ## -------------Комплексные параметры ------------------------------------------------
    # Не принесли эффекта, метрика падала
    #
    # #Отклонение от шаблона покупки
    # df_enh['days_overdue'] = df_enh['days_since_last_order'] - (30 / (df_enh['order_frequency_month'] + exp))
    #
    # # Отношение открытия приложения в заказы
    # df_enh['app_order_ratio'] =  df_enh['app_opens_per_week'] / (df_enh['order_frequency_week'] + exp)
    #
    # # Динамика изменения оценок
    # df_enh['rating_delta'] = df_enh['review_rating_last_1'] - df_enh['review_rating_last_10']
    ## ---------------------------------------------------------------------------------------------

    # Списки полей из исходного датафрейма, для добавления новых полей и передачи в пайплайн
    negative_cols = ['avg_order_value', 'total_spent_last_month','total_spent_last_week', 
                    'review_rating_last_1']

    numeric_zero_fill = ['order_frequency_month', 'total_spent_last_month', 'total_spent_last_week',
                        'app_opens_per_week', 'app_crashes_last_month', 'discount_usage_rate', 
                        'seasonal_menu_tried', 'notifications_enabled', 'coffee_preference_change']

       

    std_scaler_columns = ['avg_order_value', 'total_spent_last_month', 
                          'total_spent_last_week', 'review_rating_last_10',
                          'app_crashes_last_month', 'days_since_last_order',
                          'days_since_last_promo', 'app_opens_per_week', 
                          'order_frequency_month', 'discount_usage_rate']
    
    outlier_cols = ['days_since_last_order', 'total_spent_last_month', 
                    'total_spent_last_week', 'app_opens_per_week', 'days_since_last_promo']

    numeric_big_value_fill = ['days_since_last_order', 'days_since_last_promo']

    numeric_median_fill = ['avg_order_value', 'review_rating_last_10', 'review_rating_last_1']

    ohe_columns = ['last_coffee_type','preferred_roast',  
                        'milk_preference','coffee_bean_origin',  
                        'last_drink_size', 'subscription_status', 
                        'seasons', 'phone_type']

    te_columns = ['geo_location']

    categorical_unknow_fill = ['last_coffee_type', 'preferred_roast', 'milk_preference',
                               'coffee_bean_origin', 'last_drink_size', 'phone_type',
                               'geo_location']

    categorical_none_fill = ['subscription_status']

    categorical_mode_fill = ['seasons'] 
    

    # Поля для sqrt 
    sqrt_cols = ['total_spent_last_month', 'total_spent_last_week', 
                'app_opens_per_week', 'avg_order_value']

    # Поля для squared
    squared_cols = ['review_rating_last_10', 'discount_usage_rate','seasonal_menu_tried',
                    'order_frequency_month','app_crashes_last_month','days_since_last_promo']

    

    # Создаём новые признаки и добавляем в список для пайплайна
    # Заполняем пропуски, применяем масштабирование
    for col in sqrt_cols:
        # Квадратный корень 
        X[f'{col}_sqrt'] =  np.sqrt(np.abs(X[col]))
        numeric_zero_fill.append(f'{col}_sqrt')
        std_scaler_columns.append(f'{col}_sqrt')   

    for col in squared_cols:    
        X[f'{col}_squared'] = X[col] ** 2
        numeric_zero_fill.append(f'{col}_squared')
        std_scaler_columns.append(f'{col}_squared')

    return ( 
            X.copy(), 
            numeric_zero_fill, 
            negative_cols, 
            std_scaler_columns, 
            outlier_cols, 
            numeric_big_value_fill, 
            numeric_median_fill,
            ohe_columns,
            te_columns,
            categorical_unknow_fill,
            categorical_none_fill,
            categorical_mode_fill
            )
````

Новые признаки создаются прямым изменением `X` вне пайплайна (функция `create_features`). В ноутбуке это работает, но есть два нюанса на будущее:

2. **Продакшн.** Логику FE придётся вручную повторять для каждого нового батча данных и помнить порядок шагов.

Стандартное решение - кастомный трансформер (`BaseEstimator`, `TransformerMixin`) первым шагом пайплайна:

```python
class FeatureEngineer(BaseEstimator, TransformerMixin):

        return self          # нечего обучать

        X = X.copy()
        for col in self.sqrt_cols:
            X[f"{col}_sqrt"] = np.sqrt(X[col].clip(lower=0))
        ...
        return X
```

Тогда вся логика инкапсулирована: `pipeline.fit(X_train, y_train)` - и всё применяется само.

### <a id='toc2_6_1_'></a>[Подбор гиперпараметров](#toc0_)

В качечтве гиперпараметров для эксперемента будут использоваться:
- Коэффициент регуляризации C - определяет силу штрафа, который накладывается за слишком большие коэффициенты.
- Вид регуляризации (L1 и L2)

````python
# Функция записи метрик в датафрейм
def metrics_write(model_result,  C, penalty, class_weight):

    result =[]
   
    for key, values in model_result.items():
        if key.startswith('test_'):
            metric_name = key.replace('test_', '')         
            result_row = {
                    'metric_name':metric_name,
                    'C':C,
                    'penalty':penalty,
                    'class_weight': class_weight,
                    'value':values.mean().round(4),
                    'std':values.std().round(4)
            }
            result.append(result_row)
    return pd.DataFrame(result)
````

````python
# Коэффициент регуляризации
c_list = [0.001, 0.01, 0.1, 0.5, 1, 2, 3, 5, 10]

# Вид регуляризации
penalty_list = ['l1', 'l2']

# Веса класса
class_weights=['balanced', None]

# Промежуточный результат
all_results = []

for i, c in enumerate(c_list): 
    for penalty in penalty_list:
        for class_weight in  class_weights:
            # Признаки уже обработаны  X_processed_feature_drop     
            # pipeline_logreg_enh = Pipeline(
            #         steps=[
            #         ('imputation', create_imputation_pipeline(numeric_zero_fill=numeric_zero_fill, negative_cols=negative_cols)),        
            #         ('scaling', create_scaling_pipeline(std_scaler_columns=std_scaler_columns)),
            #         ('encoding', create_encoding_pipeline()),
            #         ('logreg', LogisticRegression(max_iter=1000, random_state=RANDOM_STATE, 
            #                                       C=c, penalty=penalty, solver='liblinear', class_weight=class_weight))
            #     ]
            # )
            logreg = LogisticRegression(max_iter=1000, random_state=RANDOM_STATE, 
                                        C=c, penalty=penalty, solver='liblinear', class_weight=class_weight)

            logreg_cv_results_eh = cross_validate(logreg, X_processed_feature_drop, y_train, cv=kfold, scoring=scoring)                
            all_results.append(metrics_write(logreg_cv_results_eh, c, penalty, class_weight))

# Датафрейм с результатами подбора гиперпараметров
df_result = pd.concat(all_results, ignore_index=True)

 
````
